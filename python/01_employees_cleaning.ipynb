# Employee Data Cleaning & Preparation

This notebook performs data inspection, cleaning, missing value handling,
duplicate removal, date transformation, and feature creation for employee data.

```python
# Import Pandas

import pandas as pd
```

```python
# Dataset Load

df = pd.read_csv("Employee_MIS_Dataset/employees.csv")
df
```

```python
# Missing Value Analysis
```

```python
# Missing Values count

df.isnull().sum()
```

```python
# Missing values percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
# Duplicate Analysis
```

```python
print(df.duplicated().sum())
```

```python
# Display duplicate rows

df[df.duplicated()]
```

```python
# Remove duplicates

df = df.drop_duplicates()
```

```python
# Verify duplicates

df.duplicated().sum()
```

```python
# Category Analysis
```

```python
# Unique department

df["Department"].unique()
```

```python
# Department wise count

df["Department"].value_counts()
```

```python
# Location wise count

df["Location"].value_counts(dropna = False)
```

```python
# Salary Analysis
```

```python
df["Salary"].describe()
```

```python
# Salary Median

print(df["Salary"].median())
```

```python
# Store Median

salary_median = df["Salary"].median()
```

```python
# Fill missing values

df["Salary"] = df["Salary"].fillna(salary_median)
```

```python
# Check Salary missing values

df["Salary"].isnull().sum()
```

```python
# Location Cleaning
```

```python
# Fill missing values

df["Location"] = df["Location"].fillna("Unknown")
```

```python
df["Location"].isnull().sum()
```

```python
# Date Handling
```

```python
# Check Date Datatype

print(df["Joining_Date"].dtype)
```

```python
# Convert Joining_date to datetime

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

```python
# Joining Year

df["Joining_Year"] = df["Joining_Date"].dt.year
```

```python
# Joining Month

df["Joining_Month"] = df["Joining_Date"].dt.month
```

```python
# Joining Month Name

df["Joining_Month_Name"] = df["Joining_Date"].dt.month_name()
```

```python
# Check Date related columns

df[[
    "Joining_Date",
    "Joining_Year",
    "Joining_Month",
    "Joining_Month_Name"
]].head()
```

```python
# Save cleaned Dataset

df.to_csv("employees_cleaned.csv",index = False)
```
