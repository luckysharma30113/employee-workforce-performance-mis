status standardization, date transformation, and leave data validation.

```python
# Duplicates
```

```python
print(df.duplicated().sum())
```

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
# Leave type
```

```python
df["Leave_Type"].unique()
```

```python
df["Leave_Type"].value_counts(dropna = False)
```

```python
df["Leave_Type"] = df["Leave_Type"].fillna("Unknown")
```

```python
df["Leave_Type"].value_counts()
```

```python
# Approval status
```

```python
df["Approval_Status"].unique()
```

```python
df["Approval_Status"].value_counts()
```

```python
df["Approval_Status"] = df["Approval_Status"].str.strip()
```

```python
# Leave Date
```

```python
print(df["Leave_Date"].dtype)
```

```python
df["Leave_Date"] = pd.to_datetime(df["Leave_Date"])
```

```python
print(df["Leave_Date"].dtype)
```

```python
# MIS Date columns
```

```python
df["Leave_Year"] = df["Leave_Date"].dt.year
```

```python
df["Leave_Month"] = df["Leave_Date"].dt.month
```

```python
df["Leave_Month_Name"] = df["Leave_Date"].dt.month_name()
```

```python
df[[
    "Leave_Date",
    "Leave_Year",
    "Leave_Month",
    "Leave_Month_Name"
]].head()
```
