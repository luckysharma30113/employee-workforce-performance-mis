# Attendance Data Cleaning & Preparation

This notebook performs data inspection, duplicate removal, status standardization,
date transformation, missing value treatment, and working-hours validation.

```python
# Duplicates Records
```

```python
# Count duplicates

print(df.duplicated().sum())
```

```python
# remove duplicates

df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
# Status analysis
```

```python
# Unique status value

df["Status"].unique()
```

```python
# Status Frequency

df["Status"].value_counts(dropna = False)
```

```python
# Remove unwnated spaces

df["Status"] = df["Status"].str.strip()
```

```python
df["Status"].unique()
```

```python
df["Status"].value_counts()
```

```python
# Date Cleaning
```

```python
# Check Date Datatype

print(df["Date"].dtype)
```

```python
# Convert Date

df["Date"] = pd.to_datetime(df["Date"])
```

```python
# Verify

print(df["Date"].dtype)
```

```python
# Date Based MIS columns
```

```python
# Year

df["Year"] = df["Date"].dt.year
```

```python
# Month Number

df["Month"] = df["Date"].dt.month
```

```python
# Month name

df["Month_Name"] = df["Date"].dt.month_name()
```

```python
# Day Name

df["Day_Name"] = df["Date"].dt.day_name()
```

```python
# Weekday/Weekend

df["Day_Type"] = df["Date"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)
```

```python
# Working Hours Analysis
```

```python
# Check missing working hours

print(df["Working_Hours"].isnull().sum())
```

```python
# Working Hours Statistics

df["Working_Hours"].describe()
```

```python
# Check Unusual working hours

df[df["Working_Hours"] > 12]
```

```python
# Wroking Hours Cleaning
```

```python
# Agar company policy ke according 12 hours se zyada valid nahi hai, to unhe missing mark kar sakte hain

df.loc[df["Working_Hours"] > 12, "Working_Hours"] = pd.NA
```

```python
df["Working_Hours"].isnull().sum()
```

```python
# Working Hours Fill
```

```python
df.loc[df["Status"].isin(["Absent","Leave"]),"Working_Hours"] = 0
```

```python
print(df["Working_Hours"].isnull().sum())
```

```python
# Baaki missing values ko median se fill kar sakte hain

working_hours_median = df["Working_Hours"].median()
```

```python
df["Working_Hours"] = df["Working_Hours"].fillna(working_hours_median)
```

```python
print(df["Working_Hours"].isnull().sum())
```
