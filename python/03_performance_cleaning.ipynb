This notebook performs data inspection, duplicate removal, status standardization,
date transformation, missing value treatment, and working-hours validation.

```python
df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

```python
# Review Month
```

```python
print(df["Review_Month"].dtype)
```

```python
df["Review_Month"] = pd.to_datetime(df["Review_Month"])
```

```python
print(df["Review_Month"].dtype)
```

```python
df["Review_Year"] = df["Review_Month"].dt.year
```

```python
df["Review_Month_Number"] = df["Review_Month"].dt.month
```

```python
df["Review_Month_Name"] = df["Review_Month"].dt.month_name()
```

```python
df[[
    "Review_Month",
    "Review_Year",
    "Review_Month_Number",
    "Review_Month_Name"
]].head()
```

```python
# Performance score
```

```python
df["Performance_Score"].describe()
```

```python
print(df["Performance_Score"].isnull().sum())
```

```python
print(df["Performance_Score"].median())
```

```python
performance_median = df["Performance_Score"].median()
```

```python
df["Performance_Score"] = df["Performance_Score"].fillna(performance_median)
```

```python
df.isnull().sum()
```

```python
# Productivity score
```

```python
df["Productivity_Score"].describe()
```

```python
df[
    (df["Productivity_Score"] < 0) |
    (df["Productivity_Score"] > 100) 
]
```

```python
df.loc[
    (df["Productivity_Score"] < 0) |
    (df["Productivity_Score"] > 100),
    "Productivity_Score"
] = pd.NA
```

```python
print(df["Productivity_Score"].isnull().sum())
```

```python
productivity_median = df["Productivity_Score"].median()
```

```python
df["Productivity_Score"] = df["Productivity_Score"].fillna(productivity_median)
```

```python
print(df["Productivity_Score"].isnull().sum())
```
